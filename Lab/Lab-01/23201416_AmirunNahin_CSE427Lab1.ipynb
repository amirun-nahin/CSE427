### Task-03

```python
a = a * 5
b = b * 4

print(a)
print(b)
```

### Task-04

```python
b = b.T

c = np.dot(a, b)
print(c)
```

# Matplotlib Library Tasks

```python
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]

plt.scatter(humidity_dhaka, temperature_dhaka, color="red", marker="*")
plt.xlabel("humidity_dhaka")
plt.ylabel("temperature_dhaka")
plt.show()
```

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(12, 8))

temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]
ax1.scatter(humidity_dhaka, temperature_dhaka, color="red", marker="*")
ax1.set(xlabel="humidity_dhaka", ylabel="temperature_dhaka")

study_hours = [2,3,4,4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
ax2.plot(marks, study_hours)
ax2.set(xlabel="marks", ylabel="study_hours")

subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
marks = [89, 90, 45, 78, 99]
ax3.barh(subjects, marks)
ax3.set(xlabel="subjects", ylabel="marks")

subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
marks = [89, 90, 45, 78, 99]
ax4.bar(subjects, marks, color=['red', 'green', 'blue', 'cyan', 'purple'])
ax4.set(xlabel="subjects", ylabel="marks")

plt.show()
```
